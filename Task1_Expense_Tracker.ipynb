# Expense Tracker (CSV/JSON)

### Objective
Build a Python-based Expense Tracker that allows users to record daily expenses, save them in CSV/JSON format, and view expense summaries.

## Importing Required Libraries

The project uses Python libraries such as CSV, JSON, and OS to manage expense data and files.

```python
import csv
import json
import os
print("Libraries imported successfully!")
```

## File Setup

CSV and JSON files are created to store and manage the expense records.

```python
csv_file = "expenses.csv"
json_file = "expenses.json"
print("Files are ready!")
```

## Add Expense

This function collects the date, category, and amount from the user and stores the expense record in the CSV and JSON files.

```python
def add_expense():
    date = input("Enter Date (DD-MM-YYYY): ")
    category = input("Enter Category (Food, Travel, Shopping, etc.): ")
    amount = float(input("Enter Amount: "))
    expense = [date, category, amount]
    # Save to CSV
    with open(csv_file, "a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(expense)
    # Save to JSON
    if os.path.exists(json_file):
        with open(json_file, "r") as file:
            data = json.load(file)
    else:
        data = []
    data.append({"Date": date, "Category": category, "Amount": amount })
    with open(json_file, "w") as file:
        json.dump(data, file, indent=4)
    print("Expense added successfully!")
```

```python
add_expense()
```

## View Expenses

This function reads the stored expense records from the CSV file and displays them to the user.

```python
def view_expenses():
    if not os.path.exists(csv_file):
        print("No expenses found!")
        return
    with open(csv_file, "r") as file:
        reader = csv.reader(file)
        print("\n----- Expense Records -----")
        for row in reader:
            print(row)
```

```python
view_expenses()
```

## Calculate Total Expense

This function calculates the total amount spent by reading all expense records and adding their amounts together.

```python
def total_expense():
    if not os.path.exists(csv_file):
        print("No expenses found!")
        return
    total = 0
    with open(csv_file, "r") as file:
        reader = csv.reader(file)
        for row in reader:
            total += float(row[2])
    print(f"Total Expense: ₹{total}")
```

```python
total_expense()
```

## Remove Expense

This function allows the user to select and remove an existing expense record from the stored data.

```python
def remove_expense():
    if not os.path.exists(csv_file):
        print("No expenses found!")
        return
    with open(csv_file, "r") as file:
        expenses = list(csv.reader(file))
    print("\nExpense List:")
    for i, expense in enumerate(expenses):
        print(f"{i + 1}. {expense}")
    index = int(input("Enter expense number to delete: ")) - 1
    if 0 <= index < len(expenses):
        deleted = expenses.pop(index)
        with open(csv_file, "w", newline="") as file:
            writer = csv.writer(file)
            writer.writerows(expenses)
        print("Deleted:", deleted)
    else:
        print("Invalid expense number!")
```

## Search Expense

This function allows the user to search for expenses based on their category and displays the matching records.

```python
def search_expense():
    if not os.path.exists(csv_file):
        print("No expenses found!")
        return
    category = input("Enter category to search: ").lower()
    found = False
    with open(csv_file, "r") as file:
        reader = csv.reader(file)
        print("\nSearch Results:")
        for row in reader:
            if row[1].lower() == category:
                print(row)
                found = True
    if not found:
        print("No matching expenses found.")
```

```python
search_expense()
```

## Menu Driven Expense Tracker

The menu-driven program allows users to add, view, calculate, remove, and search expenses through simple menu options.

```python
while True:
    print("\n====== Expense Tracker ======")
    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Total Expense")
    print("4. Delete Expense")
    print("5. Search Expense")
    print("6. Exit")
    choice = input("Enter your choice: ")
    if choice == "1":
        add_expense()
    elif choice == "2":
        view_expenses()
    elif choice == "3":
        total_expense()
    elif choice == "4":
        remove_expense()
    elif choice == "5":
        search_expense()
    elif choice == "6":
        print("Thank you for using Expense Tracker!")
        break
    else:
        print("Invalid choice! Please try again.")
```

## Conclusion

The Expense Tracker project successfully provides a simple way to manage daily expenses. Users can add, view, calculate, remove, and search expense records. The data is stored in both CSV and JSON formats, making the application useful for basic expense management and data handling.
